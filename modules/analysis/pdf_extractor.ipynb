```python
import pdfplumber
import pandas as pd

def extract_frac_tables(pdf_path):
    extracted_data = []
    
    # Open the FRAC PDF
    with pdfplumber.open(pdf_path) as pdf:
        # Loop through every page
        for i, page in enumerate(pdf.pages):
            # Extract tables from the page
            tables = page.extract_tables()
            
            for table in tables:
                # Add each row of the table to our data list
                for row in table:
                    # Clean out empty rows or heavily merged header artifacts
                    if any(cell and str(cell).strip() for cell in row):
                        normalized_row = row[:7] + [None] * max(0, 7 - len(row))
                        extracted_data.append(normalized_row)
                        
    return extracted_data

# Run the extraction
pdf_file = "frac-code-list-2026.pdf"
raw_table_data = extract_frac_tables(pdf_file)

df = pd.DataFrame(raw_table_data, columns=[
    "Mode of Action", 
    "Target Site", 
    "Group Name", 
    "Chemical Group", 
    "Common Name", 
    "FRAC Code", 
    "Comments" # Resistance Risk data will likely be found inside this column or the Group Name column
])

df.to_csv("frac_raw_data.csv", index=False)
print(df.head(200))


```
