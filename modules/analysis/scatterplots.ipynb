This is for learning to generate plots

```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#set seaborn background theme
sns.set_theme(style="whitegrid", context="talk")
#Load clean .csv files
df = pd.read_csv("rdkit_72pH_HEAT.csv")
# Create a dictionary mapping categories to Hex Color Codes
poster_colors = {
    "Decoy": "#5B646A",                
    "No-Inhibition": "#5DADE2",        
    "Native Ligand": "#B72F2F",        
    "Tophit low affinity": "#5BAE7B",  
    "Tophit high affinity": "#194C2E"  
}
# Make a square plot
fig, ax = plt.subplots(figsize=(8, 8))

# Draw the scatterplot using Seaborn
sns.scatterplot(
    data=df, 
    x='2IW3', 
    y='7B7D', 
    hue='Category', 
    s=150, 
    alpha=0.8, 
    edgecolor='k', 
    palette=poster_colors, 
    ax=ax
)
#Invert axes just by setting limits in reverse order
ax.set_xlim(-2, -8)
ax.set_ylim(-2, -8)
#Draw xy line
ax.plot([-2, -8], [-2, -8], linestyle='--', color='gray', label='No Change')
#Make sure grid stays square
ax.set_aspect('equal', adjustable='box')
#Labels
ax.set_xlabel('2IW3 (kcal/mol)')
ax.set_ylabel('7B7D (kcal/mol)')
ax.set_title('Binding Affinity HEAT pocket')
# Legend at the Bottom Center
# bbox_to_anchor=(0.5, -0.15) pushes it below the X-axis, dead center.
# ncol=2 or 3 will spread the categories out horizontally instead of a tall list.
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=False)
# Ensure nothing gets cut off when saving
plt.tight_layout()
# Save as a scalable vector graphic
plt.savefig("2IW3_vs_7B7D_HEAT_rdkit.png", format="png", bbox_inches="tight")
plt.show()
```
